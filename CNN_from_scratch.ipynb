```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import keras
```

```python
x = np.array([
    [1,2,3,0],
    [0,1,2,3],
    [3,1,0,2],
    [2,0,1,1]
])

f = np.array([
    [1 ,0],
    [0, -1]
])
```

$$
\text{out}[i,j] = \sum_{m=0}^{k_H-1} \sum_{n=0}^{k_W-1} x[i \cdot s + m,\; j \cdot s + n] \cdot w[m,n]
$$

```python
def conv2d(x, w, stride=1):
    k_h , k_w = w.shape
    out_h = x.shape[0] - k_h // 1 + 1
    out_w = x.shape[1] - k_w // 1 + 1
    out = np.zeros((out_h, out_w))

    for i in range(out_h):
        for j in range(out_w):
            patch = x[i*stride: i*stride+k_h, j*stride:j*stride+k_w]
            out[i, j] = np.sum(patch * w)
    return out

convolved = conv2d(x, f)
```

$$
\text{ReLU}(z) = \max(0, z)
$$

```python
def ReLU(x):
    return np.maximum(0 ,x)

convolution_output = ReLU(convolved)
```

$$
\text{out} = \left\lfloor \frac{\text{in} - k + 2p}{s} \right\rfloor + 1
$$

```python
def feature_map_size_calc(x_size , stride, filter_size, padding):
    out_h = (x_size[0] - filter_size[0]) + 2 * padding // stride[0] + 1
    out_w = (x_size[1] - filter_size[1]) + 2 * padding // stride[1] + 1
    return (out_h, out_w)

print(feature_map_size_calc((4,4), (1,1), (2,2), 0))
```

```python
def max_pooling(feature_maps , pool_size = (2,2)):
    fm_hs = feature_maps.shape[0]
    fm_ws = feature_maps.shape[1]

    out_h = (fm_hs - pool_size[0]) // 1 + 1
    out_w = (fm_ws - pool_size[1]) // 1 + 1

    out = np.zeros((out_h, out_w, feature_maps.shape[2]))

    for c in range(out.shape[2]):
        for i in range(out.shape[0]):
            for j in range(out.shape[1]):
                patch = feature_maps[i*1:i*1+pool_size[0] ,  
                                     j*1:j*1+pool_size[1],c]
                out[i, j , c] = np.max(patch)
    return out


fms = np.random.random((3,3,3))

pooled = max_pooling(fms)
pooled.shape
```
